Materialize canonical v1, v2, and v3 catalogue and identity products from the shared corrected ingestion pipeline.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "scripts":
    ROOT = ROOT.parent

def run_module(module, *args):
    command = [sys.executable, "-m", module, *args]
    print("+", " ".join(command))
    subprocess.run(command, cwd=ROOT, check=True)
```

```python
run_module("src.internal.process_catalogues")
```
