# 03 - Generate paper figures

Render the common paper-figure suite for v1, v2, and v3.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "scripts":
    ROOT = ROOT.parent

def run_module(module, *args):
    command = [sys.executable, "-m", module, *args]
    print("+", " ".join(command))
    subprocess.run(command, cwd=ROOT, check=True)
```

```python
run_module("src.internal.generate_figures", "v1")
run_module("src.internal.generate_figures", "v2")
run_module("src.internal.generate_figures", "v3")
```
