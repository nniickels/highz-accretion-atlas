# 02 - Generate science tables

Generate the shared point-ranking, Monte Carlo, duty-cycle, sensitivity, exclusion, and summary tables for every dataset version.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "scripts":
    ROOT = ROOT.parent

def run_module(module, *args):
    command = [sys.executable, "-m", module, *args]
    print("+", " ".join(command))
    subprocess.run(command, cwd=ROOT, check=True)
```

```python
run_module("src.internal.generate_science", "v1", "v2", "v3")
```
