# 01 - Generate science tables

Generate growth requirements, mass errors, duty cycles, and publication tables. Also build the four [extended v3 tables](../docs/reference/catalogue-results.md): compact object results, assumption comparisons, field availability, and minimum seeds.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "scripts":
    ROOT = ROOT.parent

def run_module(module, *args):
    command = [sys.executable, "-m", module, *args]
    print("+", " ".join(command))
    subprocess.run(command, cwd=ROOT, check=True)
```

```python
run_module("src.internal.generate_science", "v1", "v2", "v3")
```
