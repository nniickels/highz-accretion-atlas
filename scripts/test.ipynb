```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_context("talk")
sns.set_style("whitegrid")

processed_data_dir = r"C:\Users\nicol\OneDrive\coding\projects\Dayal\highz-accretion-atlas\data\processed\v1_processed.csv"

# Standardized catalogue
df = pd.read_csv(processed_data_dir)
```

```python
plot_df = df.dropna(subset=["redshift", "log_mbh_mstar_ratio"]).copy()

fig, ax = plt.subplots(figsize=(9, 5))
sc = ax.scatter(
    plot_df["redshift"],
    plot_df["log_mbh_mstar_ratio"],
    c=plot_df.get("edd_ratio_std", np.nan),
    cmap="viridis",
    s=70,
    alpha=0.85,
    edgecolor="k",
    linewidth=0.4,
)

ax.set_xlabel("Redshift z")
ax.set_ylabel(r"$\log_{10}(M_{\rm BH}/M_\star)$")

if "edd_ratio_std" in plot_df.columns:
    cbar = plt.colorbar(sc, ax=ax, pad=0.01)
    cbar.set_label(r" $f_{\rm Edd}$")

plt.tight_layout()
plt.show()
```
