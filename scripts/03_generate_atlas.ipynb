# 04 - Generate all-object atlases

Regenerate the flattened `fedd_mass_maps` and `seedredshift_mass_maps` galleries plus the complete f_Edd-mass, compatibility, growth-track, and uncertainty atlases.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "scripts":
    ROOT = ROOT.parent

def run_module(module, *args):
    command = [sys.executable, "-m", module, *args]
    print("+", " ".join(command))
    subprocess.run(command, cwd=ROOT, check=True)
```

```python
run_module("src.internal.generate_atlas", "v1", "--rebuild-panels")
run_module("src.internal.generate_atlas", "v2", "--rebuild-panels")
run_module("src.internal.generate_atlas", "v3", "--rebuild-panels")
run_module("src.internal.build_results_inventory")
run_module("src.internal.verify_regenerated_artifacts", "--refresh-manifests")
```
