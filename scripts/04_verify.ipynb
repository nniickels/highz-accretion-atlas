# 05 - Verify repository

Run source-provenance, canonical dataset/result, and regression verification after generation.


```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "scripts":
    ROOT = ROOT.parent

def run_module(module, *args):
    command = [sys.executable, "-m", module, *args]
    print("+", " ".join(command))
    subprocess.run(command, cwd=ROOT, check=True)
```

```python
run_module("src.internal.verify_manual_extractions")
run_module("src.internal.verify_primary_source_values")
run_module("src.internal.verify_source_provenance")
run_module("src.internal.verify_versions")
run_module("unittest", "discover", "-s", "tests")
```
